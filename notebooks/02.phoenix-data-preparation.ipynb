# PHOENIX Synthetic Spectrum Preparation

## Project Objective

This notebook prepares synthetic stellar spectra from the PHOENIX stellar atmosphere library for stellar parameter estimation.

The PHOENIX library provides pre-computed synthetic spectra for different stellar atmospheric parameters:

- Effective temperature (Teff)
- Surface gravity (log g)
- Metallicity ([Fe/H])

These synthetic spectra will later be compared with observed stellar spectra to determine the best-fitting atmospheric parameters.

## Reference

Husser et al. (2013)

"A new extensive library of PHOENIX stellar atmospheres and synthetic spectra"
Astronomy & Astrophysics, 553, A6.

# PHOENIX Library Overview

The PHOENIX library contains synthetic stellar spectra generated using stellar atmosphere models.

Each spectrum corresponds to a specific set of atmospheric parameters:

- Teff: Effective temperature
- log g: Surface gravity
- [Fe/H]&#58; Metallicity

The project uses PHOENIX spectra as theoretical models.

The goal is to compare these models with an observed spectrum around the Hβ absorption line (4861 Å).
