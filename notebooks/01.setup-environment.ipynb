# 5. Version Control Configuration (.gitignore)

A `.gitignore` file was created to prevent unnecessary files from being tracked by Git.

Python virtual environments contain thousands of dependency files and should not be included in version control because:

- They are machine-specific.
- They make the repository unnecessarily large.
- They can be recreated using `requirements.txt`.

The following files were excluded:

- Virtual environments
- Python cache files
- Jupyter temporary files
- VS Code settings

Example `.gitignore`:

```text
.venv/
__pycache__/
*.pyc
.ipynb_checkpoints/
.vscode/
```

````python
import sys
import os

print("=" * 60)
print("AST402 MidTerm Project Environment Verification")
print("=" * 60)

# Python version
print("\nPython Version:")
print(sys.version)

# Python executable
print("\nPython Executable:")
print(sys.executable)

# Check virtual environment
print("\nVirtual Environment Check:")
if ".venv" in sys.executable:
    print("✓ Running inside project virtual environment")
else:
    print("✗ Virtual environment not detected")

# Required packages
print("\nPackage Verification:")

packages = [
    "numpy",
    "scipy",
    "matplotlib",
    "pandas",
    "astropy",
    "astroquery",
    "specutils",
    "jupyter",
    "ipykernel"
]

failed_packages = []

for package in packages:
    try:
        __import__(package)
        print(f"✓ {package}")
    except ImportError:
        print(f"✗ {package}")
        failed_packages.append(package)

# Project directory check
print("\nProject Structure Check:")

directories = [
    "data",
    "data/PHOENIX",
    "notebooks",
    "scripts",
    "plots",
    "docs"
]

for directory in directories:
    if os.path.exists(directory):
        print(f"✓ {directory}/")
    else:
        print(f"✗ {directory}/")

# Final result
print("\n" + "=" * 60)

if len(failed_packages) == 0:
    print("SUCCESS: AST402 environment is ready!")
else:
    print("WARNING: Missing packages:")
    print(failed_packages)

print("=" * 60)
````
